```python
df = pd.read_csv('lettuce.csv', encoding='latin1')
df.head()
```

```python
import numpy as np
import pandas as pd

# Mempersiapkan fitur dan target dari dataset lettuce
# Fitur: Temperature (°C), Humidity (%), TDS Value (ppm), pH Level
# Target: Growth Days
X_raw = df[['Temperature (°C)', 'Humidity (%)', 'TDS Value (ppm)', 'pH Level']].values
y_raw = df[['Growth Days']].values

# Normalisasi sederhana (Min-Max Scaling agar berkisar antara 0 - 1)
X = (X_raw - X_raw.min(axis=0)) / (X_raw.max(axis=0) - X_raw.min(axis=0))
y = (y_raw - y_raw.min(axis=0)) / (y_raw.max(axis=0) - y_raw.min(axis=0))

# Parameter Jaringan Saraf Tiruan
input_size = 4
hidden_size = 4
output_size = 1
lr = 0.05

# Inisialisasi bobot
np.random.seed(42)
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Proses Training
for epoch in range(5000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1 / len(X)
    b1 += lr * d_b1 / len(X)
    W2 += lr * d_W2 / len(X)
    b2 += lr * d_b2 / len(X)

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.6f}")

# Output akhir (Denormalisasi untuk melihat beberapa hasil prediksi)
loss_akhir = np.mean(np.square(y - a2))
print(f"Loss akhir: {loss_akhir:.6f}")

prediksi_denorm = a2 * (y_raw.max() - y_raw.min()) + y_raw.min()
print("\nBeberapa hasil prediksi Growth Days vs Aktual:")
for i in range(10):
    print(f"Aktual: {y_raw[i][0]} | Prediksi: {prediksi_denorm[i][0]:.1f}")
```

```python
# Tugas 1: JST Lettuce dengan hidden layer 2 vs 3 neuron, Sigmoid vs ReLU
import numpy as np
import pandas as pd

# Mempersiapkan data lettuce
X_raw = df[['Temperature (°C)', 'Humidity (%)', 'TDS Value (ppm)', 'pH Level']].values
y_raw = df[['Growth Days']].values

# Normalisasi Min-Max
X = (X_raw - X_raw.min(axis=0)) / (X_raw.max(axis=0) - X_raw.min(axis=0))
y = (y_raw - y_raw.min(axis=0)) / (y_raw.max(axis=0) - y_raw.min(axis=0))

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def train_lettuce(hidden_size, act="sigmoid", lr=0.05, epochs=3000, seed=42):
    np.random.seed(seed)
    input_size = 4
    output_size = 1

    W1 = np.random.randn(input_size, hidden_size)
    b1 = np.zeros((1, hidden_size))
    W2 = np.random.randn(hidden_size, output_size)
    b2 = np.zeros((1, output_size))

    for epoch in range(epochs):
        # Forward pass
        z1 = np.dot(X, W1) + b1
        a1 = sigmoid(z1) if act == "sigmoid" else relu(z1)
        z2 = np.dot(a1, W2) + b2
        a2 = sigmoid(z2)

        # Hitung error
        error = y - a2

        # Backpropagation
        d_a2 = error * a2 * (1 - a2)
        d_W2 = np.dot(a1.T, d_a2)
        d_b2 = np.sum(d_a2, axis=0, keepdims=True)

        d1 = a1 * (1 - a1) if act == "sigmoid" else (z1 > 0).astype(float)
        d_a1 = np.dot(d_a2, W2.T) * d1
        d_W1 = np.dot(X.T, d_a1)
        d_b1 = np.sum(d_a1, axis=0, keepdims=True)

        # Update weights
        W1 += lr * d_W1 / len(X)
        b1 += lr * d_b1 / len(X)
        W2 += lr * d_W2 / len(X)
        b2 += lr * d_b2 / len(X)

    loss = np.mean(np.square(y - a2))
    return loss

# Bandingkan konfigurasi sesuai instruksi Tugas 1
for h, act in [(4, "sigmoid"), (3, "sigmoid"), (3, "relu")]:
    loss_val = train_lettuce(hidden_size=h, act=act)
    print(f"Hidden={h}, Aktivasi={act:8s} | Loss akhir: {loss_val:.6f}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf

# Mempersiapkan fitur dan target dari dataset lettuce
X_lettuce = df[['Temperature (°C)', 'Humidity (%)', 'TDS Value (ppm)', 'pH Level']].values
y_lettuce = df[['Growth Days']].values

# Normalisasi Standard Scaling untuk target regresi
scaler_X = StandardScaler()
scaler_y = StandardScaler()
X_scaled = scaler_X.fit_transform(X_lettuce)
y_scaled = scaler_y.fit_transform(y_lettuce)

# Split data menjadi training (80%) dan testing (20%)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_scaled, test_size=0.2, random_state=42)

# Bangun model JST Regresi
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    tf.keras.layers.Dense(10, activation='relu'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1) # Output 1 unit untuk regresi (Growth Days)
])

# Kompilasi dengan loss Mean Squared Error (MSE)
model.compile(optimizer='adam', loss='mse', metrics=['mae'])

# Latih model
history = model.fit(X_train, y_train, epochs=50, batch_size=16, validation_split=0.1, verbose=1)

# Evaluasi model pada data uji
loss, mae = model.evaluate(X_test, y_test, verbose=0)
print(f"\nHasil Evaluasi Data Uji - MSE Loss: {loss:.4f}, MAE: {mae:.4f}")
```

```python
# Tugas 2: Variasi jumlah neuron hidden layer pada Lettuce menggunakan TensorFlow/Keras
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Mempersiapkan fitur dan target
X_lettuce = df[['Temperature (°C)', 'Humidity (%)', 'TDS Value (ppm)', 'pH Level']].values
y_lettuce = df[['Growth Days']].values

# Normalisasi Standard Scaling untuk stabilitas JST Keras
scaler_X = StandardScaler()
scaler_y = StandardScaler()
X_scaled = scaler_X.fit_transform(X_lettuce)
y_scaled = scaler_y.fit_transform(y_lettuce)

# Split data
X_tr, X_te, y_tr, y_te = train_test_split(X_scaled, y_scaled, test_size=0.2, random_state=42)

def run_lettuce_keras(hidden, act="relu", epochs=50):
    tf.keras.utils.set_random_seed(42)
    layers = [tf.keras.layers.Input(shape=(4,))]
    layers += [tf.keras.layers.Dense(n, activation=act) for n in hidden]
    layers += [tf.keras.layers.Dense(1)] # Regresi output
    m = tf.keras.Sequential(layers)
    m.compile(optimizer="adam", loss="mse")
    m.fit(X_tr, y_tr, epochs=epochs, batch_size=16, verbose=0)
    return m.evaluate(X_te, y_te, verbose=0)

print("Evaluasi variasi hidden layer (MSE pada Test Set):")
for hidden in [(10, 8), (4,), (32, 16), (64, 32)]:
    mse_loss = run_lettuce_keras(hidden)
    print(f"Hidden {str(hidden):16s} -> MSE Loss: {mse_loss:.4f}")
```

```python
# Tugas 3: Bandingkan Sigmoid vs ReLU pada dataset Lettuce
print("Perbandingan Fungsi Aktivasi (Sigmoid vs ReLU) pada dataset Lettuce:")
for act in ["sigmoid", "relu"]:
    mse_loss = run_lettuce_keras((10, 8), act=act)
    print(f"{act:8s} -> MSE Loss akhir: {mse_loss:.4f}")
```

# TUGAS PRAKTIKUM
Langkah:

Load dataset MNIST dari Keras.

Bangun model dengan 2 hidden layer.

Latih model dan evaluasi akurasi.

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Bangun model JST
model = Sequential([
    Flatten(input_shape=(28, 28)),    # Ubah gambar 28x28 menjadi vektor
    Dense(128, activation='relu'),    # Hidden layer 1
    Dense(64, activation='relu'),     # Hidden layer 2
    Dense(10, activation='softmax')   # Output layer (10 kelas)
])

# 3. Kompilasi model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# 4. Latih model
model.fit(X_train, y_train, epochs=10, batch_size=32, validation_split=0.1)

# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi pada data uji: {acc:.4f}")
```

Coba dengan beberapa parameter lain:

Ubah jumlah neuron di hidden layer (misal: 256 dan 128).

Tambahkan satu hidden layer lagi.

Bandingkan akurasi dan waktu pelatihan.

Eksperimen dengan fungsi aktivasi Sigmoid vs ReLU.

```python
import time
def run_mnist(hidden, act="relu", epochs=5):
    tf.keras.utils.set_random_seed(42)
    m = Sequential([Flatten(input_shape=(28,28))] +
                   [Dense(n, activation=act) for n in hidden] +
                   [Dense(10, activation='softmax')])
    m.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    t = time.time()
    m.fit(X_train, y_train, epochs=epochs, batch_size=32, verbose=0)
    dur = time.time() - t
    l, a = m.evaluate(X_test, y_test, verbose=0)
    print(f"Hidden {str(hidden):16s} {act:8s} | Loss: {l:.4f} | Akurasi: {a:.4f} | Waktu: {dur:.1f}s")

run_mnist((128, 64))          # baseline
run_mnist((256, 128))         # neuron lebih banyak
run_mnist((256, 128, 64))     # tambah 1 hidden layer
run_mnist((128, 64), "sigmoid")  # Sigmoid vs ReLU
```
